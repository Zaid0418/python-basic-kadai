```python
df = pd.read_csv("熊本支店.csv",encoding="cp932")
```

```python
areas = ["熊本","大阪","福岡"]
df_all = pd.DataFrame()
for area in areas:
    df = pd.read_csv(f"{area}支店.csv",encoding="cp932")
    df_all = pd.concat([df_all, df])
df_all["担当部署名"].value_counts()
```

```python
df_all["請求月"] =pd.to_datetime( df_all["請求日"]).dt.to_period("M")
df_all
```

```python
df_all.groupby("請求月")["請求金額"].sum().reset_index()
```
