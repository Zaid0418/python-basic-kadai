```python
areas = ["熊本","大阪","福岡"]
df_all = pd.DataFrame()
for area in areas:
    df = pd.read_csv(f"{area}支店.csv",encoding="cp932")
    df_all = pd.concat([df_all, df])
df_all["担当部署名"].value_counts()
```

```python
df_all["請求月"] =pd.to_datetime( df_all["請求日"]).dt.to_period("M")
df_all
```

```python
jittsseki=df_all.groupby("請求月")["請求金額"].sum().reset_index()
jittsseki
```

```python
import pandas as pd
yosan = pd.read_excel("yosan_15.xlsx")
yosan
```

```python
yosan["予算月"] =pd.to_datetime( yosan["月"]).dt.to_period("M")
yosan
```

```python
# 結合（merge）
merged_df = pd.merge(jittsseki, yosan, left_on='請求月', right_on='予算月', how='left')

# 結果表示
merged_df = merged_df.drop("月", axis=1)
```

```python
merged_df["月間予算不足額"] = merged_df["月間予算"] - merged_df["請求金額"]
merged_df["予算月"] = merged_df["予算月"].astype(str)
json_str = merged_df[["予算月","月間予算不足額"]].to_json(orient='records', force_ascii=False)  # force_ascii=False で日本語をそのまま出力
json_str
```

```python
merged_df["月間予算不足額"].sum()
```
